```python
import numpy as np
import pandas as pd

participant = DATASET_DIR / "F-01"

tags = pd.read_csv(
    participant / "tags.csv",
    header=None,
    names=["timestamp"]
)

activities = pd.read_csv(
    participant / "tags_activity.csv",
    header=None,
    names=["timestamp", "activity"]
)

# Match each tag timestamp to its activity annotation
activity_lookup = {
    round(row.timestamp, 2): row.activity
    for row in activities.itertuples()
}

tags["timestamp"] = tags["timestamp"].astype(float)
tags["activity"] = tags["timestamp"].round(2).map(activity_lookup)

tags["seconds_from_previous"] = tags["timestamp"].diff()

print("F-01 COMPLETE TAG EVENT SEQUENCE")
display(tags.round(2))

# Inspect the unmatched event
missing_time = 1672590920.28

position = (
    tags.index[
        np.isclose(tags["timestamp"], missing_time, atol=0.01)
    ][0]
)

print("\nUNMATCHED EVENT:")
display(tags.iloc[[position]].round(2))

print("\nPREVIOUS EVENT:")
display(tags.iloc[[position - 1]].round(2))

print("\nNEXT EVENT:")
display(tags.iloc[[position + 1]].round(2))
```

```python
activity_pairs = {
    "rest": ("r", "er"),
    "drawing": ("d", "ed"),
    "washing_dishes": ("wd", "ewd"),
    "wiping_floor": ("wp", "ewp"),
    "walking_fast": ("wf", "ewf"),
    "upstairs": ("us", "eus"),
    "downstairs": ("ds", "eds")
}

valid_intervals = []
excluded_intervals = []

for participant in participant_folders:

    annotations = pd.read_csv(
        participant / "tags_activity.csv",
        header=None,
        names=["timestamp", "activity"]
    ).sort_values("timestamp")

    annotations["activity"] = (
        annotations["activity"].astype(str).str.strip()
    )

    for activity, (start_code, end_code) in activity_pairs.items():

        starts = annotations.loc[
            annotations["activity"] == start_code, "timestamp"
        ].tolist()

        ends = annotations.loc[
            annotations["activity"] == end_code, "timestamp"
        ].tolist()

        if len(starts) == 1 and len(ends) == 1 and ends[0] > starts[0]:

            valid_intervals.append({
                "participant": participant.name,
                "activity": activity,
                "start": starts[0],
                "end": ends[0],
                "duration_seconds": ends[0] - starts[0]
            })

        else:
            excluded_intervals.append({
                "participant": participant.name,
                "activity": activity,
                "start_count": len(starts),
                "end_count": len(ends),
                "reason": "Missing, repeated, or invalid markers"
            })

intervals_df = pd.DataFrame(valid_intervals)
excluded_df = pd.DataFrame(excluded_intervals)

print("VALID INTERVALS:", len(intervals_df))
display(intervals_df)

print("\nEXCLUDED INTERVALS:")
display(excluded_df)

print("\nVALID INTERVALS PER ACTIVITY:")
display(intervals_df.groupby("activity").size())
```

```python
import numpy as np
import pandas as pd

window_seconds = 5
overlap_seconds = 2.5
sampling_rate_expected = 32

sensor_audit = []

for participant in participant_folders:

    acc_path = participant / "ACC.csv"

    # Read the two metadata rows
    metadata = pd.read_csv(
        acc_path,
        header=None,
        nrows=2
    )

    sensor_start = float(metadata.iloc[0, 0])
    sampling_rate = float(metadata.iloc[1, 0])

    # Read actual accelerometer measurements
    acc = pd.read_csv(
        acc_path,
        header=None,
        skiprows=2,
        names=["x", "y", "z"]
    )

    sensor_end = sensor_start + (len(acc) - 1) / sampling_rate

    participant_intervals = intervals_df[
        intervals_df["participant"] == participant.name
    ]

    available_windows = 0
    outside_intervals = 0

    for interval in participant_intervals.itertuples():

        start = max(interval.start, sensor_start)
        end = min(interval.end, sensor_end)

        if end <= start:
            outside_intervals += 1
            continue

        duration = end - start

        if duration >= window_seconds:
            count = int(
                np.floor(
                    (duration - window_seconds) / overlap_seconds
                )
            ) + 1

            available_windows += count
        else:
            outside_intervals += 1

    sensor_audit.append({
        "participant": participant.name,
        "sampling_rate": sampling_rate,
        "samples": len(acc),
        "sensor_duration_seconds": sensor_end - sensor_start,
        "valid_intervals": len(participant_intervals),
        "estimated_windows": available_windows,
        "intervals_outside_sensor": outside_intervals
    })

sensor_audit_df = pd.DataFrame(sensor_audit)

display(sensor_audit_df.round(2))

print("\nTOTAL ESTIMATED WINDOWS:",
      sensor_audit_df["estimated_windows"].sum())
```

```python
import numpy as np
import pandas as pd

from pathlib import Path

FS = 32
WINDOW_SIZE = 160
STEP_SIZE = 80

feature_rows = []

for participant in participant_folders:

    # Read accelerometer metadata
    metadata = pd.read_csv(
        participant / "ACC.csv",
        header=None,
        nrows=2
    )

    sensor_start = float(metadata.iloc[0, 0])
    sampling_rate = float(metadata.iloc[1, 0])

    if sampling_rate != FS:
        raise ValueError(
            f"Unexpected sampling rate for {participant.name}: "
            f"{sampling_rate}"
        )

    # Load sensor measurements and convert to g
    acc = pd.read_csv(
        participant / "ACC.csv",
        header=None,
        skiprows=2,
        names=["x", "y", "z"]
    ).to_numpy(dtype=float) / 64.0

    participant_intervals = intervals_df[
        intervals_df["participant"] == participant.name
    ]

    for interval in participant_intervals.itertuples():

        # Convert timestamps into sample indices
        start_idx = max(
            0,
            int(np.ceil((interval.start - sensor_start) * FS))
        )

        end_idx = min(
            len(acc),
            int(np.floor((interval.end - sensor_start) * FS))
        )

        segment = acc[start_idx:end_idx]

        # Extract fixed-size windows
        for i in range(0, len(segment) - WINDOW_SIZE + 1, STEP_SIZE):

            window = segment[i:i + WINDOW_SIZE]

            features = {}

            # Axis-wise statistical features
            for axis_idx, axis_name in enumerate(["x", "y", "z"]):

                values = window[:, axis_idx]

                features[f"{axis_name}_mean"] = np.mean(values)
                features[f"{axis_name}_std"] = np.std(values)
                features[f"{axis_name}_min"] = np.min(values)
                features[f"{axis_name}_max"] = np.max(values)
                features[f"{axis_name}_median"] = np.median(values)
                features[f"{axis_name}_rms"] = np.sqrt(
                    np.mean(values ** 2)
                )

            # Three-dimensional acceleration magnitude
            magnitude = np.linalg.norm(window, axis=1)

            features["magnitude_mean"] = np.mean(magnitude)
            features["magnitude_std"] = np.std(magnitude)
            features["magnitude_rms"] = np.sqrt(
                np.mean(magnitude ** 2)
            )

            features["activity"] = interval.activity
            features["participant"] = participant.name

            feature_rows.append(features)

# Create final feature dataset
features_df = pd.DataFrame(feature_rows)

print("DATASET SHAPE:", features_df.shape)

print("\nWINDOWS PER ACTIVITY:")
display(features_df["activity"].value_counts())

print("\nWINDOWS PER PARTICIPANT:")
display(features_df["participant"].value_counts())

print("\nMISSING VALUES:")
print(features_df.isnull().sum().sum())

display(features_df.head())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Separate features, labels, and participant groups
X = features_df.drop(columns=["activity", "participant"])
y = features_df["activity"]
groups = features_df["participant"]

# Leave one participant out per fold
logo = LeaveOneGroupOut()

all_true = []
all_pred = []
fold_results = []

for train_idx, test_idx in logo.split(X, y, groups):

    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    all_true.extend(y_test)
    all_pred.extend(predictions)

    fold_results.append({
        "held_out_participant": groups.iloc[test_idx].iloc[0],
        "test_windows": len(test_idx),
        "accuracy": accuracy_score(y_test, predictions),
        "balanced_accuracy": balanced_accuracy_score(
            y_test, predictions
        ),
        "macro_f1": f1_score(
            y_test, predictions, average="macro", zero_division=0
        )
    })

# Overall participant-independent results
print("OVERALL RESULTS")
print("Accuracy:", accuracy_score(all_true, all_pred))
print("Balanced Accuracy:", balanced_accuracy_score(all_true, all_pred))
print("Macro F1:", f1_score(
    all_true, all_pred, average="macro", zero_division=0
))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    all_true,
    all_pred,
    zero_division=0
))

print("\nRESULTS BY PARTICIPANT")
display(pd.DataFrame(fold_results).round(4))
```

```python
class_labels = sorted(y.unique())

# Confusion matrix: actual rows, predicted columns
cm = confusion_matrix(
    all_true,
    all_pred,
    labels=class_labels
)

print("CONFUSION MATRIX — COUNTS")
display(pd.DataFrame(
    cm,
    index=class_labels,
    columns=class_labels
))

# Row-normalized confusion matrix
row_totals = cm.sum(axis=1, keepdims=True)

cm_normalized = np.divide(
    cm,
    row_totals,
    out=np.zeros_like(cm, dtype=float),
    where=row_totals != 0
)

print("\nCONFUSION MATRIX — NORMALIZED")
display(pd.DataFrame(
    cm_normalized.round(3),
    index=class_labels,
    columns=class_labels
))

print("\nACTIVITY DISTRIBUTION BY PARTICIPANT")
display(pd.crosstab(
    features_df["participant"],
    features_df["activity"]
))
```

```python
from scipy.stats import skew, kurtosis

def extract_frequency_features(signal, fs=32):
    features = {}

    for axis_idx, axis_name in enumerate(["x", "y", "z"]):
        x = signal[:, axis_idx].astype(float)
        x = x - np.mean(x)

        fft_values = np.abs(np.fft.rfft(x))
        frequencies = np.fft.rfftfreq(len(x), d=1/fs)

        power = fft_values ** 2
        total_power = np.sum(power)

        features[f"{axis_name}_dominant_frequency"] = (
            frequencies[np.argmax(power[1:]) + 1]
            if len(power) > 1 else 0
        )

        features[f"{axis_name}_spectral_energy"] = total_power

        features[f"{axis_name}_spectral_entropy"] = (
            -np.sum(
                (power / total_power) *
                np.log2(power / total_power + 1e-12)
            )
            if total_power > 0 else 0
        )

    return features
```

```python
frequency_rows = []
window_labels = []

FS = 32
WINDOW_SIZE = 160
STEP_SIZE = 80

for participant_folder in sorted(DATASET_DIR.iterdir()):

    if not participant_folder.is_dir():
        continue

    participant = participant_folder.name
    acc_file = participant_folder / "ACC.csv"

    if not acc_file.exists():
        continue

    metadata = pd.read_csv(acc_file, header=None, nrows=2)

    sensor_start = float(metadata.iloc[0, 0])
    sampling_rate = float(metadata.iloc[1, 0])

    acc = (
        pd.read_csv(
            acc_file,
            header=None,
            skiprows=2,
            names=["x", "y", "z"]
        ).to_numpy(dtype=float) / 64.0
    )

    participant_intervals = intervals_df[
        intervals_df["participant"] == participant
    ]

    for _, interval in participant_intervals.iterrows():

        start_idx = max(
            0,
            int(np.ceil((interval.start - sensor_start) * FS))
        )

        end_idx = min(
            len(acc),
            int(np.floor((interval.end - sensor_start) * FS))
        )

        segment = acc[start_idx:end_idx]

        for i in range(
            0,
            len(segment) - WINDOW_SIZE + 1,
            STEP_SIZE
        ):

            window = segment[i:i + WINDOW_SIZE]

            frequency_rows.append(
                extract_frequency_features(window, fs=FS)
            )

            window_labels.append({
                "participant": participant,
                "activity": interval.activity
            })

frequency_df = pd.DataFrame(frequency_rows)
frequency_labels_df = pd.DataFrame(window_labels)

print("Frequency features:", frequency_df.shape)
print("Window labels:", frequency_labels_df.shape)
print("Original baseline:", features_df.shape)

# Verify that the recreated windows match the original baseline order.
assert len(frequency_df) == len(features_df), (
    "Window counts do not match. Do not combine the datasets yet."
)

assert frequency_labels_df["participant"].reset_index(drop=True).equals(
    features_df["participant"].reset_index(drop=True)
), "Participant order mismatch."

assert frequency_labels_df["activity"].reset_index(drop=True).equals(
    features_df["activity"].reset_index(drop=True)
), "Activity order mismatch."

# Preserve the original baseline and create a separate expanded dataset.
expanded_features_df = pd.concat(
    [
        features_df.reset_index(drop=True),
        frequency_df.reset_index(drop=True)
    ],
    axis=1
)

print("\nExpanded dataset:", expanded_features_df.shape)
print("\nNew frequency features:")
display(frequency_df.head())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report
)

X = expanded_features_df.drop(columns=["activity", "participant"])
y = expanded_features_df["activity"]
groups = expanded_features_df["participant"]

class_names = sorted(y.unique())
logo = LeaveOneGroupOut()

all_true_expanded = []
all_pred_expanded = []
expanded_fold_results = []

for train_idx, test_idx in logo.split(X, y, groups):

    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    all_true_expanded.extend(y_test)
    all_pred_expanded.extend(predictions)

    expanded_fold_results.append({
        "held_out_participant": groups.iloc[test_idx].iloc[0],
        "accuracy": accuracy_score(y_test, predictions),
        "macro_f1": f1_score(
            y_test,
            predictions,
            labels=class_names,
            average="macro",
            zero_division=0
        )
    })

print("EXPANDED MODEL — OVERALL RESULTS")
print("Accuracy:", accuracy_score(all_true_expanded, all_pred_expanded))
print("Balanced Accuracy:", balanced_accuracy_score(
    all_true_expanded, all_pred_expanded
))
print("Macro F1:", f1_score(
    all_true_expanded,
    all_pred_expanded,
    labels=class_names,
    average="macro",
    zero_division=0
))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    all_true_expanded,
    all_pred_expanded,
    labels=class_names,
    target_names=class_names,
    zero_division=0
))

print("\nRESULTS BY PARTICIPANT")
display(pd.DataFrame(expanded_fold_results).round(4))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X_baseline = features_df.drop(columns=["activity", "participant"])
y = features_df["activity"]
groups = features_df["participant"]

logo = LeaveOneGroupOut()

svm_true = []
svm_pred = []
svm_fold_results = []

for train_idx, test_idx in logo.split(X_baseline, y, groups):

    X_train = X_baseline.iloc[train_idx]
    X_test = X_baseline.iloc[test_idx]
    y_train = y.iloc[train_idx]
    y_test = y.iloc[test_idx]

    svm_model = make_pipeline(
        StandardScaler(),
        SVC(
            kernel="rbf",
            C=1.0,
            gamma="scale",
            class_weight="balanced"
        )
    )

    svm_model.fit(X_train, y_train)
    predictions = svm_model.predict(X_test)

    svm_true.extend(y_test)
    svm_pred.extend(predictions)

    svm_fold_results.append({
        "held_out_participant": groups.iloc[test_idx].iloc[0],
        "accuracy": accuracy_score(y_test, predictions),
        "balanced_accuracy": balanced_accuracy_score(y_test, predictions),
        "macro_f1": f1_score(
            y_test,
            predictions,
            labels=class_names,
            average="macro",
            zero_division=0
        )
    })

print("SVM — OVERALL RESULTS")
print("Accuracy:", accuracy_score(svm_true, svm_pred))
print("Balanced Accuracy:", balanced_accuracy_score(svm_true, svm_pred))
print("Macro F1:", f1_score(
    svm_true, svm_pred,
    labels=class_names,
    average="macro",
    zero_division=0
))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    svm_true, svm_pred,
    labels=class_names,
    target_names=class_names,
    zero_division=0
))

print("\nRESULTS BY PARTICIPANT")
display(pd.DataFrame(svm_fold_results).round(4))
```

```python
from sklearn.inspection import permutation_importance
from sklearn.metrics import make_scorer

X = features_df.drop(columns=["activity", "participant"])
y = features_df["activity"]
groups = features_df["participant"]

class_names = sorted(y.unique())

macro_f1_scorer = make_scorer(
    f1_score,
    labels=class_names,
    average="macro",
    zero_division=0
)

logo = LeaveOneGroupOut()
fold_importances = []

for train_idx, test_idx in logo.split(X, y, groups):

    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    )

    model.fit(X_train, y_train)

    result = permutation_importance(
        model,
        X_test,
        y_test,
        scoring=macro_f1_scorer,
        n_repeats=5,
        random_state=42,
        n_jobs=-1
    )

    fold_importances.append(result.importances_mean)

permutation_df = pd.DataFrame({
    "feature": X.columns,
    "mean_importance": np.mean(fold_importances, axis=0),
    "std_importance": np.std(fold_importances, axis=0)
}).sort_values("mean_importance", ascending=False)

print("TOP 15 HELD-OUT PERMUTATION FEATURES")
display(permutation_df.head(15).round(4))
```
